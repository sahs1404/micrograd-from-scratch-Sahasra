```python
xs = [
    [Value(0.0), Value(0.0)],
    [Value(0.0), Value(1.0)],
    [Value(1.0), Value(0.0)],
    [Value(1.0), Value(1.0)],
]

ys = [Value(0.0), Value(1.0), Value(1.0), Value(0.0)]

for epoch in range(30):
    ypred = [model(x) for x in xs]
    loss = Value(0.0)
    for yp, yt in zip(ypred, ys):
         diff = yp - yt
         loss = loss + diff * diff   

    for p in model.parameters():
        p.grad = 0
    loss.backward()
    for p in model.parameters():
        p.data += -0.1 * p.grad

    if epoch % 5 == 0:
        print(f"epoch {epoch}, loss = {loss.data}")
```

```python
for x in xs:
    print(model(x).data)
```
